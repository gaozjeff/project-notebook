Question 8: Solve the linear system in Python and calculate the forces in the truss members.

Question 9: Determine the mid-span deflection in the vertical direction. Compare with the results

```python
#Question 8&9

import numpy as np
import sympy as sp

# Question 8: Solve for displacements and calculate member forces

# Define the force vector

F = np.zeros(N)

F[3] = -10   # joint 2
F[5] = -10   # joint 3
F[7] = -10   # joint 4

# boundary conditions

fixed_dofs = [0, 1, 9]
free_dofs = [i for i in range(N) if i not in fixed_dofs]

# stiffness matrix
K_ff = K[np.ix_(free_dofs, free_dofs)]
F_f = F[free_dofs]

#free displacements
U_free = np.linalg.solve(K_ff, F_f)

# displacement vector
U = np.zeros(N)
for i, dof in enumerate(free_dofs):
    U[dof] = U_free[i]

print("Nodal displacements (inches):")
for i in range(N):
    print(f"  DOF {i+1}: {U[i]:.6f} in")

# Calculating member forces:
member_guides = [
    guideK1, guideK2, guideK3, guideK4, guideK5,
    guideK6, guideK7, guideK8, guideK9, guideK10,
    guideK11, guideK12, guideK13, guideK14, guideK15,
    guideK16, guideK17
]
member_T = [
    mem_1_T, mem_2_T, mem_3_T, mem_4_T, mem_5_T,
    mem_6_T, mem_7_T, mem_8_T, mem_9_T, mem_10_T,
    mem_11_T, mem_12_T, mem_13_T, mem_14_T, mem_15_T,
    mem_16_T, mem_17_T
]


print("\nMember Forces (kips):")
for m in range(17):
    guide = member_guides[m]
    T = member_T[m]
    u_local = np.array([U[guide[0]], U[guide[1]], U[guide[2]], U[guide[3]]])
    # Force = (EA/L) * [−1, 1] @ T @ u_local  (deformation in local coords)
    deformation = T @ u_local  # [u_local_i, u_local_j]
    force = (E * A / L) * (deformation[1] - deformation[0])
    print(f"  Member {m+1}: {force:.2f} kips")
```
